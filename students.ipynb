```python
# создание новой категории Attendance_Category

def categorize_attendance(arr):
    if float(arr) > 90:
        return "Excellent"
    elif 50 <= float(arr) <= 90:
        return 'Average'
    else:
        return 'Poor'

df['Attendance_Category'] = df['Attendance (%)'].apply(categorize_attendance)
df[['Student_ID','Attendance (%)','Attendance_Category']].head()
```
